It finds the image and mask folders (including nested ZIP folders), verifies matching PNG filenames, and makes the paths expected by the original lab without modifying the graded game engine.

**Important:** Upload both archives in the same prompt. Training uses `CameraRGB` images and the red channel of `CameraSeg`/`CameraMask` masks. If a folder is missing or the labels are invalid, the cell stops with a clear error instead of silently guessing.


````python
# Manual data upload: run this cell in Google Colab and choose BOTH provided ZIP files.
import os
import shutil
import zipfile
from pathlib import Path
from google.colab import files

UPLOAD_ROOT = Path("/content/cs4085_lab7_uploaded")
DATA_DIR = "/content/cs4085_lab7_ready"
UPLOAD_ROOT.mkdir(parents=True, exist_ok=True)
os.makedirs(DATA_DIR, exist_ok=True)

uploaded = files.upload()
if not uploaded:
    raise ValueError("No files uploaded. Select BOTH dataset ZIPs and run this cell again.")

# Extract each archive into its own directory to avoid colliding PNG filenames.
for archive_index, filename in enumerate(uploaded.keys()):
    source_file = Path(filename)
    if not zipfile.is_zipfile(source_file):
        print(f"Skipping non-ZIP upload: {filename}")
        continue
    destination = UPLOAD_ROOT / f"archive_{archive_index}"
    destination.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(source_file) as archive:
        # Verify archive paths stay inside their extraction directory.
        for info in archive.infolist():
            candidate = (destination / info.filename).resolve()
            if not candidate.is_relative_to(destination.resolve()):
                raise ValueError(f"Unsafe ZIP member in {filename}: {info.filename}")
        archive.extractall(destination)
    print(f"Extracted: {filename}")
del uploaded

# Find folders containing PNGs. Folder names may be nested inside either archive.
folders = []
for root, dirs, names in os.walk(UPLOAD_ROOT):
    png_names = {name for name in names if name.lower().endswith(".png")}
    if png_names:
        folders.append((Path(root), png_names))

mask_options = [(p, names) for p, names in folders
                if p.name.lower() in ("cameraseg", "cameramask")]
if not mask_options:
    raise FileNotFoundError(
        "Could not locate CameraSeg/CameraMask in the uploaded ZIPs. "
        "Check that the segmentation-mask ZIP was uploaded."
    )
mask_source, mask_names = max(mask_options, key=lambda item: len(item[1]))

rgb_options = [(p, names) for p, names in folders if p.name.lower() == "camerargb"]
if not rgb_options:
    # Some ZIP creators omit the CameraRGB parent folder. Identify RGB files by
    # matching PNG filenames, excluding the known segmentation-mask directories.
    rgb_options = [(p, names) for p, names in folders
                   if p != mask_source and p.name.lower() not in ("cameraseg", "cameramask")]
if not rgb_options:
    raise FileNotFoundError(
        "Could not locate RGB images in the uploaded ZIPs. Upload the CameraRGB ZIP."
    )

image_source, image_names = max(
    rgb_options, key=lambda item: (len(item[1] & mask_names), len(item[1]))
)
pairs_count = len(image_names & mask_names)
if pairs_count < 3:
    raise ValueError(
        f"Only {pairs_count} matching image/mask PNG filenames found. "
        "Check that BOTH correct archives are uploaded."
    )

# Keep all original filename pairings and use directory links: no extra image copies.
for link_name, src in (("CameraRGB", image_source), ("CameraSeg", mask_source)):
    link = Path(DATA_DIR) / link_name
    if link.is_symlink():
        link.unlink()
    elif link.exists():
        raise FileExistsError(f"Expected path is already a real directory: {link}")
    link.symlink_to(src, target_is_directory=True)

MASK_FOLDER = "CameraSeg"
print("RGB source:", image_source)
print("Mask source:", mask_source)
print("Matched image/mask pairs:", pairs_count)
print("DATA_DIR:", DATA_DIR)
````

### 0.6 Build the data pipeline
- read the PNG files and scale image pixels to [0, 1],
- keep only the **red channel** of the mask (the class label),
- resize both to 96 × 128 with `nearest` interpolation, so labels stay whole numbers,

````python
def process_path(image_path, mask_path):
    img = tf.io.read_file(image_path)
    img = tf.image.decode_png(img, channels=3)
    img = tf.image.convert_image_dtype(img, tf.float32)
    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=3)
    mask = mask[:, :, :1]          # the class label is stored in the red channel
    return img, mask

def preprocess(image, mask):
    input_image = tf.image.resize(image, (96, 128), method='nearest')
    input_mask  = tf.image.resize(mask,  (96, 128), method='nearest')
    return input_image, input_mask

pairs = list(zip(image_list, mask_list))
np.random.default_rng(7).shuffle(pairs)
n_val = max(1, len(pairs) // 10)
val_pairs, train_pairs = pairs[:n_val], pairs[n_val:]

def make_ds(p):
    ds = tf.data.Dataset.from_tensor_slices(([a for a, _ in p], [b for _, b in p]))
    return ds.map(process_path).map(preprocess)

train_ds, val_ds = make_ds(train_pairs), make_ds(val_pairs)
print('Training images:', len(train_pairs), '| Validation images:', len(val_pairs))
````

## 🧩 Level 1 — Shape puzzle (warm-up)

### How a U-Net works
A classification CNN ends with dense layers, which throw away *where* things are. Segmentation needs a label at every pixel, so U-Net is **fully convolutional** and shaped like a **U**:

1. **Encoder (contracting path)** — blocks of two 3×3 `same` convolutions with ReLU, then 2×2 max pooling. Each step **halves** height and width and **doubles** the filters.
2. **Bottleneck** — the deepest block, with no pooling.
3. **Decoder (expanding path)** — each step uses a **transposed convolution** to **double** height and width and **halve** the filters, concatenates the matching encoder output, then applies two 3×3 convolutions.
4. **Skip connections** — carry the encoder output at each level straight to the matching decoder level, giving back the fine detail lost by pooling.
5. **Output** — a 1×1 convolution maps the features at each pixel to **13 scores**, one per class.

```
cblock1 ──────────────skip──────────────► ublock9 → conv9 → conv10 (output)
   cblock2 ──────────skip──────────► ublock8
      cblock3 ──────skip──────► ublock7
         cblock4 ──skip──► ublock6
                 cblock5 (bottleneck)
```

### 2.2 Exercise 1 — `conv_block`
Each encoder block returns **two** tensors:
- `next_layer` — goes into the next encoder block,
- `skip_connection` — goes to the matching decoder block.

If `max_pooling=True`, `next_layer` is the pooling output and `skip_connection` is the tensor **before** pooling. Otherwise both are the same tensor.

**Implement it:**
1. Two **Conv2D** layers: `n_filters` filters, `kernel_size=3`, `activation='relu'`, `padding='same'`, `kernel_initializer='he_normal'`.
2. If `dropout_prob > 0`, a **Dropout** layer with rate `dropout_prob`.
3. If `max_pooling` is `True`, a **MaxPooling2D** layer with `pool_size=(2, 2)`.

````python
# GRADED FUNCTION: conv_block
def conv_block(inputs=None, n_filters=32, dropout_prob=0, max_pooling=True):
    """
    Convolutional downsampling block
    Arguments:
        inputs -- Input tensor
        n_filters -- Number of filters for the convolutional layers
        dropout_prob -- Dropout probability
        max_pooling -- Use MaxPooling2D to reduce the spatial dimensions of the output volume
    Returns:
        next_layer, skip_connection -- Next layer and skip connection outputs
    """
    ### START CODE HERE
    conv = Conv2D(n_filters, 3,
                  activation='relu',
                  padding='same',
                  kernel_initializer='he_normal')(inputs)
    conv = Conv2D(n_filters, 3,
                  activation='relu',
                  padding='same',
                  kernel_initializer='he_normal')(conv)
    ### END CODE HERE

    if dropout_prob > 0:
        ### START CODE HERE
        conv = Dropout(dropout_prob)(conv)
        ### END CODE HERE

    if max_pooling:
        ### START CODE HERE
        next_layer = MaxPooling2D(pool_size=(2, 2))(conv)
        ### END CODE HERE
    else:
        next_layer = conv

    skip_connection = conv
    return next_layer, skip_connection
````

### 3.1 Exercise 2 — `upsampling_block`
The decoder block has two new parts:
- `up` — a **transposed convolution** that doubles height and width,
- `merge` — the **skip connection**: `up` concatenated with the matching encoder output.

Then two Conv2D layers with the same settings as in the encoder.

**Implement it:**
1. **Conv2DTranspose** on `expansive_input`: `n_filters` filters, kernel size `3`, `strides=(2, 2)`, `padding='same'`.
2. The concatenation is written for you: `concatenate([up, contractive_input], axis=3)`. Keep this order.
3. **Two Conv2D** layers: `n_filters`, kernel size `3`, `activation='relu'`, `padding='same'`, `kernel_initializer='he_normal'`.

````python
# GRADED FUNCTION: upsampling_block
def upsampling_block(expansive_input, contractive_input, n_filters=32):
    """
    Convolutional upsampling block
    Arguments:
        expansive_input -- Input tensor from previous layer
        contractive_input -- Input tensor from previous skip layer
        n_filters -- Number of filters for the convolutional layers
    Returns:
        conv -- Tensor output
    """
    ### START CODE HERE
    up = Conv2DTranspose(
                 n_filters,
                 3,
                 strides=(2, 2),
                 padding='same')(expansive_input)

    merge = concatenate([up, contractive_input], axis=3)
    conv = Conv2D(n_filters, 3,
                  activation='relu',
                  padding='same',
                  kernel_initializer='he_normal')(merge)
    conv = Conv2D(n_filters, 3,
                  activation='relu',
                  padding='same',
                  kernel_initializer='he_normal')(conv)
    ### END CODE HERE
    return conv
````

**Decoder:**
1. `ublock6`: `expansive_input = cblock5[0]`, `contractive_input = cblock4[1]`, filters `n_filters*8`.
2. Each next block takes the previous decoder output and the **second** output `[1]` (before pooling) of the matching encoder block. **Halve** the filters each step: `*4`, `*2`, `*1`.

**Output:**
- `conv9` is given.
- `conv10`: Conv2D with `n_classes` filters, kernel size `1`, `padding='same'`, **no activation** (the loss works on raw scores).

````python
# GRADED FUNCTION: unet_model
def unet_model(input_size=(96, 128, 3), n_filters=32, n_classes=13):
    """
    Unet model
    Arguments:
        input_size -- Input shape
        n_filters -- Number of filters for the convolutional layers
        n_classes -- Number of output classes
    Returns:
        model -- tf.keras.Model
    """
    inputs = Input(input_size)
    ### START CODE HERE
    cblock1 = conv_block(inputs, n_filters)
    cblock2 = conv_block(cblock1[0], n_filters * 2)
    cblock3 = conv_block(cblock2[0], n_filters * 4)
    cblock4 = conv_block(cblock3[0], n_filters * 8, dropout_prob=0.3)
    cblock5 = conv_block(cblock4[0], n_filters * 16,
                         dropout_prob=0.3, max_pooling=False)
    ### END CODE HERE

    ### START CODE HERE
    ublock6 = upsampling_block(cblock5[0], cblock4[1], n_filters * 8)
    ublock7 = upsampling_block(ublock6, cblock3[1], n_filters * 4)
    ublock8 = upsampling_block(ublock7, cblock2[1], n_filters * 2)
    ublock9 = upsampling_block(ublock8, cblock1[1], n_filters)
    ### END CODE HERE

    conv9 = Conv2D(n_filters, 3, activation='relu', padding='same',
                   kernel_initializer='he_normal')(ublock9)

    ### START CODE HERE
    conv10 = Conv2D(n_classes, 1, padding='same')(conv9)
    ### END CODE HERE

    return tf.keras.Model(inputs=inputs, outputs=conv10)
````

````python
def display(display_list):
    plt.figure(figsize=(15, 15))
    title = ['Input Image', 'True Mask', 'Predicted Mask']
    for i in range(len(display_list)):
        plt.subplot(1, len(display_list), i + 1)
        plt.title(title[i])
        plt.imshow(tf.keras.utils.array_to_img(display_list[i]))
        plt.axis('off')
    plt.show()

def create_mask(pred_mask):
    pred_mask = tf.argmax(pred_mask, axis=-1)
    pred_mask = pred_mask[..., tf.newaxis]
    return pred_mask[0]

for image, mask in val_dataset.unbatch().batch(1).take(4):
    display([image[0], mask[0], create_mask(unet.predict(image, verbose=0))])
````
